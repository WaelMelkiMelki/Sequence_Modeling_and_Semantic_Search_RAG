```python
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

# 1. Load the pre-trained Transformer model
# This might take a few seconds to download the first time
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Define the dimension of the vectors
# This specific model outputs vectors with 384 dimensions
d = 384 

print("Libraries imported and Model loaded successfully!")
```

```python
# 1. Define the Raw Text (The Data)
raw_text = """
CRISPR-Cas9 is a unique technology that enables geneticists and medical researchers to edit parts of the genome by removing, adding or altering sections of the DNA sequence. 
It is currently the simplest, most versatile and precise method of genetic manipulation and is therefore causing a buzz in the science world. 
The CRISPR-Cas9 system consists of two key molecules that introduce a change into the DNA. 
These are: an enzyme called Cas9. 
This acts as a pair of 'molecular scissors' that can cut the two strands of DNA at a specific location in the genome so that bits of DNA can then be added or removed. 
A piece of RNA called guide RNA (gRNA). 
This consists of a small piece of pre-designed RNA sequence located within a longer RNA scaffold. 
The scaffold part binds to DNA and the pre-designed sequence 'guides' Cas9 to the right part of the genome. 
This makes sure that the Cas9 enzyme cuts at the right point in the genome.
"""

# 2. Define the Optimized Chunking Function (Sentence-based)
def create_chunks(text, window_size=3, step=1):
    """
    Splits text into overlapping groups of sentences.
    window_size: Number of sentences per chunk (e.g., 3).
    step: Slide by 1 sentence to create high overlap.
    """
    # Split by period to get sentences. 
    # We add the '.' back because split() removes it.
    sentences = [s.strip() + "." for s in text.split('.') if s.strip()]
    
    chunks = []
    
    # Create sliding windows
    # If we have 5 sentences and window=3: [1,2,3], [2,3,4], [3,4,5]
    for i in range(0, len(sentences) - window_size + 1, step):
        chunk = " ".join(sentences[i:i + window_size])
        chunks.append(chunk)
        
    return chunks

# 3. Process the data
# We group 3 sentences together. This ensures the "Answer" is complete.
sequences = create_chunks(raw_text, window_size=3, step=1)

# 4. Verification
print(f"Generated {len(sequences)} sentence-based chunks.\n")
print("--- Preview of First Chunk ---")
print(sequences[0])
```

```python
# --- RE-BUILD THE DATABASE ---

# 1. Re-generate embeddings for the NEW sentence-based sequences
embeddings = model.encode(sequences)
embeddings = np.array(embeddings).astype('float32')

# 2. Reset the FAISS Index (Clear the old data)
index = faiss.IndexFlatL2(d) 

# 3. Add the NEW vectors
index.add(embeddings)

print("Database updated!")
print(f"New Index Size: {index.ntotal}")
print(f"New Sequence List Size: {len(sequences)}")
```

```python
# 1. Initialize the FAISS Index
# We use 'd' (384) to tell FAISS how big the vectors are
index = faiss.IndexFlatL2(d) 

# 2. Add the vectors to the index
# This stores our embedding data inside the FAISS structure
index.add(embeddings)

# 3. Verification
print("Vector Database built successfully!")
print(f"Total vectors stored in FAISS: {index.ntotal}")
```

```python
def search(query, k=3):
    """
    1. Encode query to vector.
    2. Search FAISS index.
    3. Retrieve original text chunks.
    """
    # 1. Vectorize the query
    query_vector = model.encode([query])
    query_vector = np.array(query_vector).astype('float32')
    
    # 2. Search the Index
    # k = number of results to return
    # D = Distances (lower is better for L2)
    # I = Indices (the IDs of the matching chunks)
    D, I = index.search(query_vector, k)
    
    print(f"Query: '{query}'")
    print("-" * 50)
    
    # 3. Context Reconstruction
    for i, idx in enumerate(I[0]):
        matched_chunk = sequences[idx]
        distance = D[0][i]
        
        print(f"Result {i+1} (Distance: {distance:.4f}):")
        print(f"... {matched_chunk} ...\n")

# --- TEST THE SYSTEM ---

# Test 1: Asking about the mechanism
search("How does the enzyme cut the DNA?")

# Test 2: Asking about the components
search("What is the role of the RNA scaffold?")
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# 1. Prepare the data
# We combine the text embeddings with the query vector to see where they land
query_text = "How does the enzyme cut the DNA?"
query_vec = model.encode([query_text])
all_vectors = np.concatenate([embeddings, query_vec])

# 2. Reduce dimensions from 384 -> 2 using PCA
pca = PCA(n_components=2)
reduced_vectors = pca.fit_transform(all_vectors)

# 3. Plotting
plt.figure(figsize=(12, 8))

# Plot all text chunks (Blue dots)
# We exclude the last one because that is our query
plt.scatter(reduced_vectors[:-1, 0], reduced_vectors[:-1, 1], c='blue', label='Text Chunks')

# Plot the Query (Red Star)
plt.scatter(reduced_vectors[-1, 0], reduced_vectors[-1, 1], c='red', marker='*', s=200, label='User Query')

# Add labels to a few points so we can see which text is which
for i, text in enumerate(sequences):
    # Only label every 2nd chunk to avoid clutter
    if i % 2 == 0:
        plt.text(reduced_vectors[i, 0]+0.02, reduced_vectors[i, 1]+0.02, f"Chunk {i}", fontsize=9)

plt.title(f"Vector Space Visualization (Query: '{query_text}')")
plt.legend()
plt.grid(True)
plt.show()
```
